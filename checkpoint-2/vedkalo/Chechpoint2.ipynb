# Полный CSV  

```python
metadata_dir = r"C:\Users\vedma\OneDrive\Рабочий стол\Учеба\HSE project\downloaded_tracks\metadata\metadata_json"

files = [f for f in os.listdir(metadata_dir) if f.endswith(".json")]
rows = []

for file in tqdm(files):
    path = os.path.join(metadata_dir, file)
    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)
        
    row = {
        "id": data.get("id"),
        "name": data.get("name"),
        "duration": data.get("duration"),
        "artist_id": data.get("artist_id"),
        "artist_name": data.get("artist_name"),
        "album_id": data.get("album_id"),
        "album_name": data.get("album_name"),
        "releasedate": data.get("releasedate"),
    }

    mi = data.get("musicinfo", {})
    row["lang"] = mi.get("lang")
    row["vocalinstrumental"] = mi.get("vocalinstrumental")
    row["gender"] = mi.get("gender")
    row["acousticelectric"] = mi.get("acousticelectric")
    row["speed"] = mi.get("speed")

    tags = mi.get("tags", {})
    row["genres"] = ", ".join(tags.get("genres", []))
    row["instruments"] = ", ".join(tags.get("instruments", []))
    row["vartags"] = ", ".join(tags.get("vartags", []))

    st = data.get("stats", {})
    row["rate_downloads_total"] = st.get("rate_downloads_total")
    row["rate_listened_total"] = st.get("rate_listened_total")
    row["playlisted"] = st.get("playlisted")
    row["favorited"] = st.get("favorited")
    row["likes"] = st.get("likes")
    row["dislikes"] = st.get("dislikes")
    row["avgnote"] = st.get("avgnote")

    rows.append(row)

df = pd.DataFrame(rows)

out_path = r"C:\Users\vedma\OneDrive\Рабочий стол\Учеба\HSE project\downloaded_tracks\metadata\metadata_full.csv"
df.to_csv(out_path, index=False, encoding="utf-8-sig")
```

#### Подсчет количества жанров, инструментов и тегов + наличие часто встречающихся (более 100 раз)

```python
import pandas as pd

df_tags = df_encoded.copy()

df_tags["genres_count"] = df_tags["genres"].apply(
    lambda x: len(str(x).split(", ")) if pd.notnull(x) else 0
)
df_tags["instruments_count"] = df_tags["instruments"].apply(
    lambda x: len(str(x).split(", ")) if pd.notnull(x) else 0
)
df_tags["vartags_count"] = df_tags["vartags"].apply(
    lambda x: len(str(x).split(", ")) if pd.notnull(x) else 0
)

def make_binary_tags(df, col, min_count=100):
    all_tags = df[col].dropna().str.split(", ").sum()
    counts = pd.Series(all_tags).value_counts()
    frequent = counts[counts >= min_count].index
    for tag in frequent:
        tag_col = f"has_{col}_{tag.replace(' ', '_')}"
        df[tag_col] = df[col].apply(lambda x: 1 if pd.notnull(x) and tag in x else 0)
    return df

for col in ["genres", "instruments", "vartags"]:
    df_tags = make_binary_tags(df_tags, col)

new_cols = [c for c in df_tags.columns if c.endswith("_count") or c.startswith("has_")]
display(df_tags[new_cols].columns)
display(df_tags[new_cols])

out_path = r"C:\Users\vedma\OneDrive\Рабочий стол\Учеба\HSE project\downloaded_tracks\metadata\metadata_only_tag_features.csv"
df_tags[new_cols].to_csv(out_path, index=False, encoding="utf-8-sig")
```

# Метрики растояний + knn

## Косинусное растояние (все фичи)

```python
X = features_df.to_numpy().astype(float)

def cosine_distance_matrix(X):
    norms = np.linalg.norm(X, axis=1, keepdims=True)
    Xn = X / norms
    sim = Xn @ Xn.T
    return 1 - sim

cos_dist = cosine_distance_matrix(X)

k = 5
near = []
nearest_distances = []

for i in range(len(X)):
    row = cos_dist[i]
    row1 = np.argsort(row)[1:k+1]
    dists = row[row1]

    near.append(row1)
    nearest_distances.append(dists)

df_cosine = pd.DataFrame({
    "track": np.arange(len(X)),
    "n1": [x[0] for x in near],
    "n2": [x[1] for x in near],
    "n3": [x[2] for x in near],
    "n4": [x[3] for x in near],
    "n5": [x[4] for x in near],
    "d1": [x[0] for x in nearest_distances],
    "d2": [x[1] for x in nearest_distances],
    "d3": [x[2] for x in nearest_distances],
    "d4": [x[3] for x in nearest_distances],
    "d5": [x[4] for x in nearest_distances],
})

df_knn_cosine
```

## Евклидово растояние (все фичи)

```python
X = features_df.to_numpy().astype(float)

def euclidean(X):
    diff = X[:, None, :] - X[None, :, :]
    dist = np.sqrt((diff ** 2).sum(axis=2))
    return dist

eu_dist = euclidean(X)

k = 5
near = []
near_dist = []

for i in range(len(X)):
    row = eu_dist[i]
    row1 = np.argsort(row)[1:k+1]
    dists = row[row1]

    near.append(row1)
    near_dist.append(dists)

df_knn_euclidean = pd.DataFrame({
    "track": np.arange(len(X)),
    "n1": [x[0] for x in near],
    "n2": [x[1] for x in near],
    "n3": [x[2] for x in near],
    "n4": [x[3] for x in near],
    "n5": [x[4] for x in near],
    "d1": [x[0] for x in near_dist],
    "d2": [x[1] for x in near_dist],
    "d3": [x[2] for x in near_dist],
    "d4": [x[3] for x in near_dist],
    "d5": [x[4] for x in near_dist],
})

df_knn_euclidean
```

## MANHATTAN distance (все фичи)

```python
X = features_df.to_numpy().astype(float)

def manhattan(X):
    diff = np.abs(X[:, None, :] - X[None, :, :])
    dist = diff.sum(axis=2)
    return dist

man_dist = manhattan(X)

k = 5
near = []
near_dist = []

for i in range(len(X)):
    row = man_dist[i]
    row1 = np.argsort(row)[1:k+1]
    dists = row[row1]

    near.append(row1)
    near_dist.append(dists)

df_knn_manhattan = pd.DataFrame({
    "track": np.arange(len(X)),
    "n1": [x[0] for x in near],
    "n2": [x[1] for x in near],
    "n3": [x[2] for x in near],
    "n4": [x[3] for x in near],
    "n5": [x[4] for x in near],
    "d1": [x[0] for x in near_dist],
    "d2": [x[1] for x in near_dist],
    "d3": [x[2] for x in near_dist],
    "d4": [x[3] for x in near_dist],
    "d5": [x[4] for x in near_dist],
})

df_knn_manhattan
```

## Correlation distance (только для числовых фичей)

```python
num_cols = [c for c in features_df.columns if c.endswith("_scaled") or c.endswith("_count")]
X_num = features_df[num_cols].to_numpy().astype(float)

def correlation(X):
    Xc = X - X.mean(axis=1, keepdims=True)
    Xn = Xc / Xc.std(axis=1, keepdims=True)
    sim = Xn @ Xn.T / X.shape[1]
    return 1 - sim

corr_dist = correlation(X_num)

k = 5
near = []
near_dist = []

for i in range(len(X_num)):
    row = corr_dist[i]
    row1 = np.argsort(row)[1:k+1]
    dists = row[row1]

    near.append(row1)
    near_dist.append(dists)

df_knn_corr = pd.DataFrame({
    "track": np.arange(len(X_num)),
    "n1": [x[0] for x in near],
    "n2": [x[1] for x in near],
    "n3": [x[2] for x in near],
    "n4": [x[3] for x in near],
    "n5": [x[4] for x in near],
    "d1": [x[0] for x in near_dist],
    "d2": [x[1] for x in near_dist],
    "d3": [x[2] for x in near_dist],
    "d4": [x[3] for x in near_dist],
    "d5": [x[4] for x in near_dist],
})

df_knn_corr
```

## Mahalanobis distance (только для числовых фичей)

```python
S = np.cov(X_num, rowvar=False)
S_inv = np.linalg.pinv(S)

def mahalanobis(X, S_inv):
    n = X.shape[0]
    dist = np.zeros((n, n))

    for i in range(n):
        diff = X - X[i]
        dist[i] = np.sqrt(np.sum(diff @ S_inv * diff, axis=1))
    return dist

maha_dist = mahalanobis(X_num, S_inv)

k = 5
near = []
near_dist = []

for i in range(len(X_num)):
    row = maha_dist[i]
    row1 = np.argsort(row)[1:k+1]
    dists = row[row1]

    near.append(row1)
    near_dist.append(dists)

df_knn_mahalanobis = pd.DataFrame({
    "track": np.arange(len(X_num)),
    "n1": [x[0] for x in near],
    "n2": [x[1] for x in near],
    "n3": [x[2] for x in near],
    "n4": [x[3] for x in near],
    "n5": [x[4] for x in near],
    "d1": [x[0] for x in near_dist],
    "d2": [x[1] for x in near_dist],
    "d3": [x[2] for x in near_dist],
    "d4": [x[3] for x in near_dist],
    "d5": [x[4] for x in near_dist],
})

df_knn_mahalanobis
```

## Jaccard distance (тольео для бинарных фичей)

```python
bin_cols = [c for c in features_df.columns if c.startswith("has_") or c.startswith("lang_") or c.startswith("vocalinstrumental_") or c.startswith("gender_") or c.startswith("acousticelectric_") or c.startswith("speed") ]
X_bin = features_df[bin_cols].to_numpy().astype(int)

def jaccard(B):
    inter = B @ B.T
    row_sum = B.sum(axis=1, keepdims=True)
    union = row_sum + row_sum.T - inter

    sim = np.zeros_like(inter, dtype=float)
    np.divide(inter, union, out=sim, where=union != 0)

    return 1 - sim

jac_dist = jaccard(X_bin)

k = 5
near = []
near_dist = []

for i in range(len(X_bin)):
    row = jac_dist[i]
    row1 = np.argsort(row)[1:k+1]
    dists = row[row1]

    near.append(row1)
    near_dist.append(dists)

df_knn_jaccard = pd.DataFrame({
    "track": np.arange(len(X_bin)),
    "n1": [x[0] for x in near],
    "n2": [x[1] for x in near],
    "n3": [x[2] for x in near],
    "n4": [x[3] for x in near],
    "n5": [x[4] for x in near],
    "d1": [x[0] for x in near_dist],
    "d2": [x[1] for x in near_dist],
    "d3": [x[2] for x in near_dist],
    "d4": [x[3] for x in near_dist],
    "d5": [x[4] for x in near_dist],
})

df_knn_jaccard
```

# Метрики качества

## Критерий релевантности: Трек считается релевантным, если совпадает хотя бы один жанр, хотя бы один инструмент, хотя бы один язык и хотя бы один вартег.

```python
genre_cols = [c for c in features_df.columns if c.startswith("has_genres_")]
instr_cols = [c for c in features_df.columns if c.startswith("has_instruments_")]
lang_cols  = [c for c in features_df.columns if c.startswith("lang_")]
vartag_cols = [c for c in features_df.columns if c.startswith("has_vartags_")]

genre_mat = features_df[genre_cols].astype(int).values
instr_mat = features_df[instr_cols].astype(int).values
lang_mat  = features_df[lang_cols].astype(int).values
vartag_mat = features_df[vartag_cols].astype(int).values

genre_match = (genre_mat @ genre_mat.T) > 0
instr_match = (instr_mat @ instr_mat.T) > 0
lang_match  = (lang_mat  @ lang_mat.T) > 0
vartag_match = (vartag_mat @ vartag_mat.T) > 0

# критерий
rel_strict = (
    genre_match.astype(int)
    * instr_match.astype(int)
    * lang_match.astype(int)
    * vartag_match.astype(int)
)

np.fill_diagonal(rel_strict, 0)

def precision_at_k(knn_df, rel_mat, k):
    scores = []
    for i in range(len(knn_df)):
        neigh = knn_df.iloc[i][["n"+str(j) for j in range(1,k+1)]].astype(int).values
        rels = rel_mat[i, neigh]
        scores.append(np.mean(rels))
    return np.mean(scores)

def recall_at_k(knn_df, rel_mat, k):
    scores = []
    for i in range(len(knn_df)):
        neigh = knn_df.iloc[i][["n"+str(j) for j in range(1,k+1)]].astype(int).values   
        rels = rel_mat[i, neigh]
        total_rels = rel_mat[i].sum()
        if total_rels == 0:
            continue
        scores.append(rels.sum() / total_rels)
    return np.mean(scores)

def ndcg_at_k(knn_df, rel_mat, k):
    scores = []
    for i in range(len(knn_df)):
        neigh = knn_df.iloc[i][["n"+str(j) for j in range(1,k+1)]].astype(int).values
        rels = rel_mat[i, neigh]
        dcg = np.sum(rels / np.log2(np.arange(2, k+2)))
        ideal = np.sort(rel_mat[i])[::-1][:k]
        idcg = np.sum(ideal / np.log2(np.arange(2, k+2)))
        if idcg == 0:
            continue
        scores.append(dcg / idcg)
    return np.mean(scores)

knn_tables = {
    "cosine": df_knn_cosine,
    "euclidean": df_knn_euclidean,
    "manhattan": df_knn_manhattan,
    "mahalanobis": df_knn_mahalanobis,
    "correlation": df_knn_corr,
    "jaccard": df_knn_jaccard
}

results = []
for name, knn_df in knn_tables.items():
    p5 = precision_at_k(knn_df, rel_strict, k=5)
    r5 = recall_at_k(knn_df, rel_strict, k=5)
    n5 = ndcg_at_k(knn_df, rel_strict, k=5)
    results.append([name, p5, r5, n5])

df_fnl = pd.DataFrame(results, columns=["metric", "precision@5", "recall@5", "ndcg@5"])
df_fnl
```

## Критерий релевантности: Трек считается релевантным, если совпадает не меньше 60% жанров, 60% инструментов и 60% вартегов, которые есть у исходного трека, плюс совпадает язык.

```python
genre_cols  = [c for c in features_df.columns if c.startswith("has_genres_")]
instr_cols  = [c for c in features_df.columns if c.startswith("has_instruments_")]
vartag_cols = [c for c in features_df.columns if c.startswith("has_vartags_")]
lang_cols   = [c for c in features_df.columns if c.startswith("lang_")]

genre_mat  = features_df[genre_cols].astype(int).values
instr_mat  = features_df[instr_cols].astype(int).values
vartag_mat = features_df[vartag_cols].astype(int).values
lang_mat   = features_df[lang_cols].astype(int).values

genre_overlap  = genre_mat @ genre_mat.T
instr_overlap  = instr_mat @ instr_mat.T
vartag_overlap = vartag_mat @ vartag_mat.T
lang_overlap   = lang_mat @ lang_mat.T

genre_req  = np.ceil(genre_mat.sum(axis=1)  * 0.6)[:, None]
instr_req  = np.ceil(instr_mat.sum(axis=1)  * 0.6)[:, None]
vartag_req = np.ceil(vartag_mat.sum(axis=1) * 0.6)[:, None]
lang_req   = np.ceil(lang_mat.sum(axis=1)   * 0.6)[:, None]

rel_strict = (
    (genre_overlap  >= genre_req).astype(int) &
    (instr_overlap  >= instr_req).astype(int) &
    (vartag_overlap >= vartag_req).astype(int) &
    (lang_overlap   >= lang_req).astype(int)
).astype(int)

np.fill_diagonal(rel_strict, 0)

def precision_at_k(knn_df, rel_mat, k):
    scores = []
    for i in range(len(knn_df)):
        neigh = knn_df.iloc[i][["n"+str(j) for j in range(1,k+1)]].astype(int).values
        rels = rel_mat[i, neigh]
        scores.append(np.mean(rels))
    return np.mean(scores)

def recall_at_k(knn_df, rel_mat, k):
    scores = []
    for i in range(len(knn_df)):
        neigh = knn_df.iloc[i][["n"+str(j) for j in range(1,k+1)]].astype(int).values   
        rels = rel_mat[i, neigh]
        total_rels = rel_mat[i].sum()
        if total_rels == 0:
            continue
        scores.append(rels.sum() / total_rels)
    return np.mean(scores)

def ndcg_at_k(knn_df, rel_mat, k):
    scores = []
    for i in range(len(knn_df)):
        neigh = knn_df.iloc[i][["n"+str(j) for j in range(1,k+1)]].astype(int).values
        rels = rel_mat[i, neigh]
        dcg = np.sum(rels / np.log2(np.arange(2, k+2)))
        ideal = np.sort(rel_mat[i])[::-1][:k]
        idcg = np.sum(ideal / np.log2(np.arange(2, k+2)))
        if idcg == 0:
            continue
        scores.append(dcg / idcg)
    return np.mean(scores)

knn_tables = {
    "cosine":      df_knn_cosine,
    "euclidean":   df_knn_euclidean,
    "manhattan":   df_knn_manhattan,
    "mahalanobis": df_knn_mahalanobis,
    "correlation": df_knn_corr,
    "jaccard":     df_knn_jaccard
}

results = []
for name, knn_df in knn_tables.items():
    p5 = precision_at_k(knn_df, rel_strict, k=5)
    r5 = recall_at_k(knn_df, rel_strict, k=5)
    n5 = ndcg_at_k(knn_df, rel_strict, k=5)
    results.append([name, p5, r5, n5])

df_fnl = pd.DataFrame(results, columns=["metric", "precision@5", "recall@5", "ndcg@5"])
df_fnl
```
